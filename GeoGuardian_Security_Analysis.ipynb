```python

import geobench
from pathlib import Path

# Find your downloaded dataset
folder = Path("data/classification_v1.0").resolve()

print("Using folder:", folder)

# Load the official benchmark task
tasks = geobench.task_iterator(
    benchmark_dir=str(folder)
)

eurosat = next(
    task for task in tasks
    if task.dataset_name == "m-eurosat"
)

# Only open training and validation data
train_data = eurosat.get_dataset(split="train")
val_data = eurosat.get_dataset(split="val")

print("SUCCESS! Real GEO-Bench data loaded!")
print("Training images:", len(train_data))
print("Validation images:", len(val_data))

sample = train_data[0]

print("First image label:", sample.label)
print("Number of bands:", len(sample.bands))
print("First band shape:", sample.bands[0].data.shape)
```

```python
val_data = eurosat.get_dataset(split="valid")

print("Training images:", len(train_data))
print("Validation images:", len(val_data))
print("First label:", train_data[0].label)
```

```python

import torch
import torch.nn as nn
import numpy as np
import json
import time
from pathlib import Path
from torch.utils.data import Dataset, DataLoader

# Use the RASID GPU when available
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using:", device)

# Convert GEO-Bench samples into PyTorch tensors
class SatelliteData(Dataset):
    def __init__(self, source):
        self.source = source

    def __len__(self):
        return len(self.source)

    def __getitem__(self, index):
        sample = self.source[index]

        # Keep the original 13-channel order
        image = np.stack(
            [band.data for band in sample.bands]
        ).astype("float32")

        image = np.nan_to_num(image)
        image = np.clip(image, 0, 10000) / 10000.0

        label = int(sample.label)

        return (
            torch.from_numpy(image),
            torch.tensor(label, dtype=torch.long)
        )

train_loader = DataLoader(
    SatelliteData(train_data),
    batch_size=32,
    shuffle=True,
    num_workers=0
)

val_loader = DataLoader(
    SatelliteData(val_data),
    batch_size=32,
    shuffle=False,
    num_workers=0
)

# Small 13-channel image classifier
class GeoModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.network = nn.Sequential(
            nn.Conv2d(13, 16, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Conv2d(16, 32, 3, padding=1),
            nn.ReLU(),
            nn.AdaptiveAvgPool2d(1),
            nn.Flatten(),
            nn.Linear(32, 10)
        )

    def forward(self, x):
        return self.network(x)

model_geo = GeoModel().to(device)

optimizer = torch.optim.Adam(
    model_geo.parameters(), lr=0.001
)
loss_fn = nn.CrossEntropyLoss()

start = time.time()

# One genuine training epoch
model_geo.train()
total_loss = 0

for images, labels in train_loader:
    images = images.to(device)
    labels = labels.to(device)

    optimizer.zero_grad()
    outputs = model_geo(images)
    loss = loss_fn(outputs, labels)
    loss.backward()
    optimizer.step()

    total_loss += loss.item()

print("Training finished!")

# Evaluate on validation only
model_geo.eval()
correct = 0
total = 0

with torch.no_grad():
    for images, labels in val_loader:
        images = images.to(device)
        labels = labels.to(device)

        predictions = model_geo(images).argmax(dim=1)

        correct += (predictions == labels).sum().item()
        total += labels.size(0)

accuracy = correct / total
elapsed = time.time() - start

print("Real validation accuracy:", round(accuracy * 100, 2), "%")
print("Elapsed seconds:", round(elapsed, 1))

# Save the trained model
torch.save(model_geo.state_dict(), "geoguardian_eurosat.pt")

# Save actual experiment evidence
result = {
    "dataset": "m-eurosat",
    "epochs": 1,
    "learning_rate": 0.001,
    "validation_accuracy": accuracy,
    "elapsed_seconds": elapsed
}

Path("geoguardian_results.json").write_text(
    json.dumps(result, indent=2)
)

print("Model and results saved!")
```

```python

import torch
import json
from pathlib import Path

# Load the best saved checkpoint available
checkpoint = Path("geoguardian_best.pt")

if not checkpoint.exists():
    checkpoint = Path("geoguardian_eurosat.pt")

model_geo.load_state_dict(
    torch.load(
        checkpoint,
        map_location=device,
        weights_only=True
    )
)

model_geo.eval()

def check_reliability(loader, noise_strength=0.02):
    clean_correct = 0
    noisy_correct = 0
    total = 0

    with torch.no_grad():
        for images, labels in loader:
            images = images.to(device)
            labels = labels.to(device)

            # Normal satellite images
            clean_predictions = model_geo(images).argmax(1)

            # Add controlled Gaussian noise
            noisy_images = torch.clamp(
                images + torch.randn_like(images) * noise_strength,
                0, 1
            )

            noisy_predictions = model_geo(noisy_images).argmax(1)

            clean_correct += (clean_predictions == labels).sum().item()
            noisy_correct += (noisy_predictions == labels).sum().item()
            total += labels.size(0)

    return {
        "clean_accuracy": 100 * clean_correct / total,
        "noisy_accuracy": 100 * noisy_correct / total
    }

results = check_reliability(val_loader)

drop = results["clean_accuracy"] - results["noisy_accuracy"]

print("GEOGUARDIAN RELIABILITY REPORT")
print("Checkpoint:", checkpoint)
print(f"Clean accuracy: {results['clean_accuracy']:.2f}%")
print(f"Noisy accuracy: {results['noisy_accuracy']:.2f}%")
print(f"Performance drop: {drop:.2f} percentage points")

Path("geoguardian_security_report.json").write_text(
    json.dumps({
        "checkpoint": str(checkpoint),
        "noise_strength": 0.02,
        **results,
        "performance_drop_points": drop
    }, indent=2)
)

print("Reliability report saved!")
```

```python

import torch
import json
from pathlib import Path

# Load the best model you have saved
model_geo.load_state_dict(
    torch.load(
        "geoguardian_best.pt",
        map_location=device,
        weights_only=True
    )
)

model_geo.eval()

# Security-inspired stress tests
tests = {
    "clean": 0.0,
    "mild_noise": 0.02,
    "stronger_noise": 0.08
}

report = {}

with torch.no_grad():
    for test_name, noise_strength in tests.items():
        correct = 0
        total = 0
        per_class_correct = torch.zeros(10)
        per_class_total = torch.zeros(10)

        for images, labels in val_loader:
            images = images.to(device)
            labels = labels.to(device)

            if noise_strength > 0:
                images = torch.clamp(
                    images + torch.randn_like(images) * noise_strength,
                    0, 1
                )

            predictions = model_geo(images).argmax(1)

            correct += (predictions == labels).sum().item()
            total += labels.numel()

            for c in range(10):
                mask = labels == c
                per_class_total[c] += mask.sum().item()
                per_class_correct[c] += (
                    (predictions == labels) & mask
                ).sum().item()

        accuracy = 100 * correct / total

        report[test_name] = {
            "accuracy": round(accuracy, 2),
            "per_class_accuracy": {
                str(c): round(
                    100 * per_class_correct[c].item()
                    / per_class_total[c].item(), 2
                )
                if per_class_total[c] > 0 else None
                for c in range(10)
            }
        }

        print(f"{test_name}: {accuracy:.2f}%")

Path("geoguardian_extended_report.json").write_text(
    json.dumps(report, indent=2)
)

print("Extended security report saved!")
```
